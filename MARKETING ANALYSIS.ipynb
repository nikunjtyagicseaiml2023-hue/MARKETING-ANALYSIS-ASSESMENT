```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("marketing_campaign_dataset 2.csv")
df
```

```python
campaign_analysis = df.groupby("Campaign_Type")["Impressions"].agg(
    ["sum", "mean", "count"]
).sort_values("sum", ascending=False)

campaign_analysis
```

```python
channel_analysis = df.groupby("Channel_Used")["Impressions"].agg(
    ["sum", "mean", "count"]
).sort_values("sum", ascending=False)

channel_analysis
```

```python
location_analysis = df.groupby("Location")["Impressions"].sum().sort_values(
    ascending=False
)

location_analysis
```

```python
total_impressions = df["Impressions"].sum()
average_impressions = df["Impressions"].mean()

print("Total Impressions:", total_impressions)
print("Average Impressions:", average_impressions)
```

```python
channel_analysis["sum"].plot(kind="bar", figsize=(8,5))

plt.title("Total Impressions by Channel")
plt.xlabel("Channel")
plt.ylabel("Total Impressions")
plt.xticks(rotation=45)
plt.show()
```
